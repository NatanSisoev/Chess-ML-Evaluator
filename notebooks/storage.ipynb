```python
from chess_eval import *
from sklearn.ensemble import GradientBoostingRegressor
```

```python
fds = StorageManager()
fds.list_datasets()
```

```python
dm = DataManager(
    filepath=DATASET_FILE,
    read_size=100_000,
    sample_size=10_000,
    transformers=[PieceInfo],
)
```

```python
fds.save_dataset(dm, "123456789")
```

```python
fds.list_datasets()
```

```python
dm_read, meta = fds.load_dataset("123456789")
dm_read.df
```

```python
model = GradientBoostingRegressor(
    loss="huber",
    learning_rate=0.03,
    n_estimators=1000,
    max_depth=5,
    subsample=0.8,
    max_features="sqrt",
    random_state=42
)

mm = ModelManager(model, dm)
mm.fit()
mm.predict()
```

```python
fds.list_models()
```

```python
fds.save_model(mm, "123456789")
```
