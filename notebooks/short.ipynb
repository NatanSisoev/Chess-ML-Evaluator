# Minimal Working Example

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
from constants import *
from managers import DataManager, ModelManager, MetricsManager
from features import FEATURE_TRANSFORMERS
```

```python
dm = DataManager(
    filepath=DATASET_FILE,
    read_size=1_000_000,
    sample_size=100_000,
    transformers=FEATURE_TRANSFORMERS,
)
```

```python
model = GradientBoostingRegressor(
    loss="huber",
    learning_rate=0.03,
    n_estimators=1000,
    max_depth=5,
    subsample=0.8,
    max_features="sqrt",
    random_state=42
)

mm_gb = ModelManager(model, dm)
```

```python
mm_gb.fit()
mm_gb.predict()
```

```python
metm = MetricsManager(mm_gb)
metm.report()
metm.plot_tol_acc()
metm.plot_scatter()
```
