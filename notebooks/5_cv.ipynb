# 5. Cross-Validation

## 5.1 Problem Setting

In the first notebook `1_eda.ipynb`, we saw in Section 1.5 that the data is not independent, so consecutive rows are correlated since they all belong to the same game. The solution suggested in the following Section 1.6 is to sample games by groups or chunks of 100 games (which is the approximate size of positions observed in the corresponding graphs). This comes into play when actually splitting the data into train and test datasets, which is correctly done using cross-validation.

Very conveniently, the `sklearn` library already has a cross-validation class `GroupKFold` which implements the desired behavior.

## 5.2 Method

Since cross-validation has to do with the training, predicting and testing of the models, we will include `cross_validate` method to our `ModelManager` class which, given a `DataManager` and other parameters implements the cross-validation and outputs the results, including the metric and time for fitting and predicting.

One parameter we want to be able to tweak is `game_len`, which we will leave at the default value of 100 as predicted previously. We will also pass the scoring metric and number of folds.

Another problem which may arise is unbalanced representation in each fold. If we split the data by chunks of 100 (or supposed games), it can happen that entire games which are biased towards the white or black pieces end up in the same folds. A possible solution would be to define a stratified version for our problem, by choosing a custom class for each group and then splitting accordingly. An alternative and rather simple solution would be to also group games and suppose that since the entire dataset is balanced, each game group will also be balanced.
