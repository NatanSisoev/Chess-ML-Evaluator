# Save DataFrame with Features

```python
from chess_eval import *
sm = StorageManager()
```

## `no_downcast_one`: 1 million (read_size=1_000_000, sample_size=100_000) with all features (82) with **no downcast**

```python
df_name = "no_downcast_one"
if df_name not in sm.list_datasets()["Name"].values:
    no_downcast_one_dm = DataManager(  # ~1 minute
        read_size=1_000_000,
        sample_size=100_000,
        downcast=False
    )
    if sm.save_dataset(no_downcast_one_dm, df_name):
        del no_downcast_one_dm
else:
    print(f"Dataframe '{df_name}' already exists.")
del df_name
```

## `one`: 1 million (read_size=1_000_000, sample_size=100_000) with all features (82)

```python
df_name = "one"
if df_name not in sm.list_datasets()["Name"].values:
    one_dm = DataManager(  # ~1 minute
        read_size=1_000_000,
        sample_size=100_000,
    )
    if sm.save_dataset(one_dm, df_name):
        del one_dm
else:
    print(f"Dataframe '{df_name}' already exists.")
del df_name
```

## `one_full`: 1 million full (read_size=1_000_000, frac=1) with all features (82)

```python
df_name = "one_full"
if df_name not in sm.list_datasets()["Name"].values:
    one_full_dm = DataManager(  # ~10 minutes
        read_size=1_000_000,
        frac=1,
    )
    if sm.save_dataset(one_full_dm, df_name):
        del one_full_dm
else:
    print(f"Dataframe '{df_name}' already exists.")
del df_name
```

## Save all

```python
from tqdm import tqdm

datasets = {
    "no_downcast_one": {
        "read_size": 1_000_000,
        "sample_size": 100_000,
        "downcast": False,
    },
    "one": {
        "read_size": 1_000_000,
        "sample_size": 100_000,
    },
    "one_full": {
        "read_size": 1_000_000,
        "frac": 1,
    },
    "one_full_rd": {
        "read_size": None,
        "sample_size": 1_000_000,
    },
    "four_full_rd": {
        "read_size": None,
        "sample_size": 4_000_000,
    },
}

for df_name, params in tqdm(datasets.items(), desc="Processing datasets", leave=False):
    if df_name in sm.list_datasets()["Name"].values:
        print(f"Dataframe '{df_name}' already exists.")
        continue

    dm = DataManager(**params)
    if sm.save_dataset(dm, df_name):
        del dm
```
