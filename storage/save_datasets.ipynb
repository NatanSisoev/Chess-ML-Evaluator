# Save DataFrame with Features

```python
from chess_eval import *
sm = StorageManager()
```

## `one`: 1 million (read_size=1_000_000, sample_size=100_000) with all features (82)

```python
one_dm = DataManager(  # ~2 minutes
    read_size=1_000_000,
    sample_size=100_000,
    transformers=FEATURE_TRANSFORMERS,
)
sm.save_dataset(one_dm, "one")
```

## `one_full`: 1 million full (read_size=1_000_000, frac=1) with all features (82)
