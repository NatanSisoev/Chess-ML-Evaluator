# Optimized And Fitted Models Storage

```python
from chess_eval import *
from inspect import signature

sm = StorageManager()
```

```python
one_dm = load_dataset("one")
one_full_dm = load_dataset("one_full")
one_full_rd_dm = load_dataset("one_full_rd")

dms = {
    "one": one_dm,
    "one_full": one_full_dm,
    "one_full_rd": one_full_rd_dm,
}

with tqdm(OPTIMAL_MODELS.items(), desc="Saving", leave=False) as pbar:  # ~3 minutes 20 seconds
    for model_name, model in pbar:
        if model_name == "KNeighborsRegressor": continue  # fitted knn doesn't make sense, it's just the dataframe itself
        pbar.set_postfix({"model": model_name})
        for dm_name, dm in dms.items():
            pbar.set_postfix({"model": model_name, "dataframe": dm_name})

            if "random_state" in signature(model).parameters:
                mm = ModelManager(model(**OPTIMAL_PARAMS[model_name], random_state=RANDOM_STATE))
            else:
                mm = ModelManager(model(**OPTIMAL_PARAMS[model_name]))

            mm.fit(dm.X_train, dm.y_train)
            sm.save_model(mm, dm, f"{ABBREVIATIONS[model_name]}_{dm_name}")
```

```python
four_full_rd_dm = load_dataset("four_full_rd")

with tqdm(OPTIMAL_MODELS.items(), desc="Saving", leave=False) as pbar:  # ~7 minutes 50 seconds
    for model_name, model in pbar:
        if model_name == "KNeighborsRegressor": continue
        pbar.set_postfix({"model": model_name, "dataframe": "four_full_rd"})

        if "random_state" in signature(model).parameters:
            mm = ModelManager(model(**OPTIMAL_PARAMS[model_name], random_state=RANDOM_STATE))
        else:
            mm = ModelManager(model(**OPTIMAL_PARAMS[model_name]))

        mm.fit(four_full_rd_dm.X_train, four_full_rd_dm.y_train)
        sm.save_model(mm, four_full_rd_dm, f"{ABBREVIATIONS[model_name]}_four_full_rd")
```
