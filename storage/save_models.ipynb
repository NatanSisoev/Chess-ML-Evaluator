# Optimized And Fitted Models Storage

```python
from chess_eval import *
from inspect import signature
sm = StorageManager()
```

```python
dataset_names = ["one", "one_full", "one_full_rd", "four_full_rd", "one_full_rd_skip", "four_full_rd_skip"]

for dm_name in tqdm(dataset_names, desc="Datasets", leave=False):  # ~10 minutes
    dm = load_dataset(dm_name)

    for model_name, model in OPTIMAL_MODELS.items():
        if model_name == "KNeighborsRegressor":
            continue  # skip, just the dataframe itself

        model_tag = f"{ABBREVIATIONS[model_name]}_{dm_name}"
        if model_tag in sm.list_models()["Name"].values:
            print(f"Model '{model_tag}' already exists.")
            continue  # skip if already saved

        if "random_state" in signature(model).parameters:
            mm = ModelManager(model(**OPTIMAL_PARAMS[model_name], random_state=RANDOM_STATE))
        else:
            mm = ModelManager(model(**OPTIMAL_PARAMS[model_name]))

        mm.fit(dm.X_train, dm.y_train)
        sm.save_model(mm, dm, model_tag)
```
